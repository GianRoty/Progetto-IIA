## Analisi esplorativa dei dati


Data in input una lista di documenti, vengono calcolate le medie statistiche del numero di parole e frasi contenute in tali documenti. 

Tale approccio viene utilizzato per determinare il numero ottimale di parole chiave da cercare.

```python
import statistics

def explorative_analysis(documents: list[str]) -> tuple[float, float]:
  count_word_in_sentences = []
  count_sentence_in_docs = []

  for doc in documents:
    num_sen = len([sen for sen in doc.split('.')])
    count_sentence_in_docs.append(num_sen)
    for sen in doc.split('.'):
      num_word = len([word for word in sen.split(' ')])
      count_word_in_sentences.append(num_word)

  return [statistics.mean(count_word_in_sentences), statistics.mean(count_sentence_in_docs)]
```

## Preprocess task

Preprocessa il testo dato in input nel seguente modo:
1. tokenizza il testo in parole
2. converte le parole in minuscolo
3. rimuove la punteggiatura e i caratteri non alfanumerici
4. lemmatizza ogni parola

```python
import nltk

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('punkt_tab')
```

```python
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

stop_words = set(stopwords.words('italian'))
lemmatizer = WordNetLemmatizer()

# Preprocesses the text given in input
def preprocess_text(text: str) -> str:
    tokens = word_tokenize(text, language='italian')
    cleaned_tokens = [
        lemmatizer.lemmatize(word.lower())
        for word in tokens
        if word.isalnum() and word.lower() not in stop_words and word not in string.punctuation
    ]

    return " ".join(cleaned_tokens)
```

Viene applicato il preprocessing a tutti i dati presenti nella colonna specificata del dataframe fornito.

```python
def get_cleaned_documents(column: str, dataframe):
  return [preprocess_text(doc) for doc in dataframe[column]]
```

# Keyword extraction


Viene effettuata l'estrazione delle parole chiave dei testi dati in input mediante l'approccio TF - IDF, in modo da non ricadere in un'analisi troppo condizionata dalla quantità ma ben bilanciata tra quantità e qualità della parola.

#### Term Frequency (TF)
Misura la frequenza di una parola in un singolo documento.

TF(t, d) = (Freqenza di t in d)/(Numero totale di termini in d)

#### Inverse Document Frequency (IDF)
Misura quanto una parola è rara nell'intero corpus. Più è rara, maggiore sarà il suo peso.

IDF(t,D) = log(1+DF(t))

#### TF-IDF Term Frequency - Inverse Document Frequency

TF−IDF(t,d,D) = TF(t,d) ⋅ IDF(t,D)

```python
from sklearn.feature_extraction.text import TfidfVectorizer

def extract_keywords_tfidf(texts, top_n=15):
    # Preprocess the text
    preprocessed_texts = [preprocess_text(text) for text in texts]

    # Vectorize the text with TF-IDF
    vectorizer = TfidfVectorizer(max_features=top_n)
    tfidf_matrix = vectorizer.fit_transform(preprocessed_texts)

    # Get the feature names (words) and sort by importance (TF-IDF score)
    feature_names = vectorizer.get_feature_names_out()
    scores = tfidf_matrix.sum(axis=0).A1

    # Create a list of (word, score) tuples
    word_scores = list(zip(feature_names, scores))
    word_scores = sorted(word_scores, key=lambda x: x[1], reverse=True)

    # Extract the top N keywords
    keywords = [(word, score) for word, score in word_scores[:top_n]]
    return keywords
```

## N-grams extraction

Genera le n-grams più frequenti da un elenco di testi utilizzando il CountVectorizer.


```python
from sklearn.feature_extraction.text import CountVectorizer

def generate_n_grams(texts, ngram = 3, top_n = 25):
  # Initialize CountVectorizer for ngrams
  vectorizer = CountVectorizer(ngram_range=(ngram, ngram), stop_words=stopwords.words('italian'))

  # Fit and transform the text
  X = vectorizer.fit_transform(texts)

  # Get n-gram feature names and their counts
  ngrams = vectorizer.get_feature_names_out()
  counts = X.toarray().sum(axis=0)

  # Sort and display the most common ngrams
  sorted_ngrams = sorted(zip(ngrams, counts), key=lambda x: x[1], reverse=True)

  return sorted_ngrams[:top_n]
```

# Compare task

Si confrontano due insiemi di parole chiave individuando
1. le parole chiave comuni a entrambi gli insiemi
2. le parole chiave uniche del primo insieme
3. le parole chiave uniche del secondo insieme

```python
def compare_words(kw1, kw2):
  # Common keywords (intersection)
  common_keywords = set(kw1.keys()) & set(kw2.keys())

  # Unique keywords in each analysis
  unique_to_1 = set(kw1.keys()) - set(kw2.keys())
  unique_to_2 = set(kw2.keys()) - set(kw1.keys())

  print(f"Common keywords: {common_keywords}")
  print(f"Unique text keywords: {unique_to_1}")
  print(f"Unique article keywords: {unique_to_2}")
```

Confronta i punteggi associati a parole chiave comuni in due dizionari e visualizza i risultati in un grafico a barre. I punteggi vengono normalizzati per garantire una migliore comparazione.

```python
def compare_score(kw1, kw2):
  common_keywords = set(kw1.keys()) & set(kw2.keys())
  
  # Check if there are common keywords to compare
  if not common_keywords:
    print("No common keywords found. Skipping comparison.")
    return
  
  # Normalize score for better visualization
  text_score_array = list(kw1.values())
  article_score_array = list(kw2.values())
  text_mean = statistics.mean(text_score_array)
  article_mean = statistics.mean(article_score_array)
  text_standard_deviation = statistics.stdev(text_score_array)
  article_standard_deviation = statistics.stdev(article_score_array)

  # Compare scores for common keywords
  score_comparison = {
    word: (abs((kw1[word] - text_mean) / text_standard_deviation), abs((kw2[word] - article_mean) / article_standard_deviation))
    for word in common_keywords
  }

  # Create DataFrame for comparison
  df = pd.DataFrame(score_comparison, index=["Text", "Article"]).T

  # Plot bar chart
  df.plot(kind="bar", figsize=(10,5), title="Keyword Score Comparison")
  plt.ylabel("TF-IDF Score")
  plt.xticks(rotation=45)
  plt.grid()
  plt.show()
```

Confronta la similarità tra due insiemi di parole chiave utilizzando la similarità del coseno basata su TF-IDF. La similarità del coseno è una misura di somiglianza tra due vettori numerici, che può essere utilizzata per confrontare documenti, set di parole o qualsiasi altro tipo dato che può essere rappresentato come vettore. Dati due vettori A e B, la similiarità del coseno viene calcolata attraverso la seguente formula

cosine_similarity(A, B) = (A * B)/ ||A||*||B||

```python
from sklearn.metrics.pairwise import cosine_similarity

def compare_similarity(kw1, kw2):
  # Convert keywords to a single string
  text_1 = " ".join(kw1.keys())
  text_2 = " ".join(kw2.keys())

  # Compute TF-IDF and similarity
  vectorizer = TfidfVectorizer()
  tfidf_matrix = vectorizer.fit_transform([text_1, text_2])
  similarity = cosine_similarity(tfidf_matrix)[0, 1]

  print(f"Keyword Similarity Score: {similarity:.2f}")
```

Viene effettuata un'analisi lessicale attraverso spacy tramite il modello "it_core_news_lg". Questo permette di ricavare i POS tags delle parole chiave che vengono raggruppati e mostrati tramite un grafico

```python
import spacy
import networkx as nx
import matplotlib.pyplot as plt

# Load the Italian spacy model
nlp = spacy.load("it_core_news_lg")

# Define a color map for POS tags
color_map = {
    "DET": "blue",      # Determiners
    "NOUN": "green",    # Nouns
    "VERB": "red",      # Verbs
    "ADV": "orange",    # Adverbs
    "ADJ": "purple",    # Adjectives
    "PRON": "pink",     # Pronouns
    "ADP": "cyan",      # Adpositions (Prepositions, Conjunctions)
    "NUM": "brown"      # Numerals
}

def lexical_analysis(kw):
    text = " ".join(sorted(kw.keys()))  # Ensure consistent word order

    # Process the text with spacy to get the POS tagging
    doc = nlp(text)

    # Create a graph
    G = nx.Graph()

    # Add nodes with word and POS tag as attributes
    for token in doc:
        G.add_node(token.text, pos=token.pos_)

    # Add edges between words with the same POS tag to form clusters
    for token1 in doc:
        for token2 in doc:
            if token1 != token2 and token1.pos_ == token2.pos_:
                G.add_edge(token1.text, token2.text)

    # Assign colors based on POS tags
    node_colors = [color_map.get(G.nodes[node]["pos"], "gray") for node in G.nodes]

    fixed_layout = nx.spring_layout(G, seed=42, k=10, iterations=300)

    # Draw the graph
    plt.figure(figsize=(12, 12))
    nx.draw(G, with_labels=True, node_color=node_colors, font_weight="bold",
            node_size=3000, font_size=12, font_color="white", edge_color="gray",
            pos=fixed_layout)  # Use fixed layout

    # Add title
    plt.title("Word Clusters by POS Tags", fontsize=16)

    # Add the legend for the POS tags and their respective colors
    handles = [plt.Line2D([0], [0], marker='o', color='w', markerfacecolor=color_map[pos], markersize=10) for pos in color_map]
    labels = [pos for pos in color_map]
    plt.legend(handles=handles, labels=labels, title="POS Tags", loc="upper left", fontsize=12)

    # Display the plot
    plt.show()
```

ANALISI SENTIMENTALE

Si svolge un'analisi sentimentale tramite il modello "nlptown/bert-base-multilingual-uncased-sentiment". Tale analisi è costituita dalle seguenti fasi:

1. Definizione del modello (pretrainato) e tokenizzazione.
2. Suddivisione del testo in chunk. Dato che la maggior parte dei documenti forniti hanno lunghezza maggiore rispetto alla massima lunghezza consentita dal modello, è stata effettuata una suddivisione in chunck, che consiste in uno splitting semantico del testo tramite il metodo sent_tokenize importato dalla libreria nltk. Se un documento, ottenuto attraverso lo split, ha una lunghezza maggiore di quella consentita, allora verranno analizzati solamente i primi 'max_length' token di tale documento. Per ottenere l'analisi completa del documento, si calcola la media dei sentimenti ottenuti dei vari chunck di tale documento. 
3. Risulta possibile visualizzare la lineplot degli score relativi all'analisi dei vari chunck per avere una visualizzazione non influenzata dalla media. 
4. Restituisce una lista di dizionari {sentiment: score} che verrà utilizzata per la visualizzazione grafica dei risultati.

(PS: Si consiglia l'esecuzione di tale modulo mediante la GPU per via dell'elevato tempo computazionale)

```python
import torch
import matplotlib.pyplot as plt
import seaborn as sns
from transformers import AutoModelForSequenceClassification, AutoTokenizer, pipeline
from collections import defaultdict
from nltk import sent_tokenize

# Load model and tokenizer
model_name = "nlptown/bert-base-multilingual-uncased-sentiment"
model = AutoModelForSequenceClassification.from_pretrained(model_name)
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Auto-detect device (GPU if available, otherwise CPU)
device = 0 if torch.cuda.is_available() else -1
batch_size = 16 if device == 0 else 4  # Adjust batch size based on device

sentiment_analyzer = pipeline('sentiment-analysis', model=model, tokenizer=tokenizer, device=device)

LABEL_MAPPING = {
    "1 star": "Very Negative",
    "2 stars": "Negative",
    "3 stars": "Neutral",
    "4 stars": "Positive",
    "5 stars": "Very Positive"
}

# Define a color palette for sentiments
SENTIMENT_COLORS = {
    "Very Negative": "red",
    "Negative": "orange",
    "Neutral": "gray",
    "Positive": "green",
    "Very Positive": "blue"
}

def analyze_sentiment_batch(texts):
    """
    Processes a batch of texts using sentiment analysis.
    """
    results = sentiment_analyzer(texts, batch_size=batch_size)
    return [{"sentiment": LABEL_MAPPING[item["label"]], "score": item["score"]} for item in results]

def process_sentiment_analysis(documents, max_length=520, generate_plots=False):
    """
    Batch-process sentiment analysis on documents with optional visualization.
    """
    all_sentences = []
    doc_mapping = []  # Track which document each sentence belongs to

    # Pre-process: Split long documents into smaller chunks
    for idx, doc in enumerate(documents):
        sentences = sent_tokenize(doc, language='italian')
        chunked_sentences = [s[:max_length] for s in sentences]  # Truncate long sentences
        all_sentences.extend(chunked_sentences)
        doc_mapping.extend([idx] * len(chunked_sentences))  # Store mapping

    # Perform sentiment analysis in batches
    sentiment_results = analyze_sentiment_batch(all_sentences)

    # Aggregate results per document
    doc_sentiments = defaultdict(lambda: {"chunks": defaultdict(list)})

    for i, result in enumerate(sentiment_results):
        doc_idx = doc_mapping[i]
        sentiment = result["sentiment"]
        score = result["score"]

        doc_sentiments[doc_idx]["chunks"][sentiment].append((i, score))  # Store chunk-level sentiment

    # Compute final sentiment per document
    final_results = []
    for idx, data in doc_sentiments.items():
        aggregated = [
            {"sentiment": sentiment, "score": np.mean([score for _, score in data["chunks"][sentiment]])}
            for sentiment in data["chunks"] if data["chunks"][sentiment]
        ]
        final_results.append(aggregated)

        # Generate plots only if `generate_plots` is True
        if generate_plots:
            # Create a 2x3 grid for subplots (3 in the first row, 2 in the second row)
            fig, axes = plt.subplots(2, 3, figsize=(18, 10), sharey=True)

            # Flatten the axes array for easier iteration
            axes = axes.flatten()

            # Plot sentiment variation across the document in 5 subplots, one for each sentiment
            for i, sentiment in enumerate(SENTIMENT_COLORS.keys()):
                ax = axes[i]  # Get the axis for the i-th sentiment plot
                chunks = data["chunks"][sentiment]
                chunk_indices = [chunk[0] for chunk in chunks]
                chunk_scores = [chunk[1] for chunk in chunks]

                if chunk_scores:  # Ensure there is data to plot
                    sns.lineplot(x=chunk_indices, y=chunk_scores, marker="o", label=sentiment, color=SENTIMENT_COLORS[sentiment], ax=ax)

                    # Calculate and plot the mean line (black line)
                    mean_score = np.mean(chunk_scores)
                    ax.axhline(mean_score, color='black', linestyle='--', label=f"{sentiment} Mean")

                    ax.set_title(f"Sentiment Trends for Document {idx+1} - {sentiment}")
                    ax.set_ylabel("Sentiment Score")
                    ax.legend()

            # Adjust x-axis for shared axes and show the plot
            plt.xlabel("Chunk Index")
            plt.xticks(rotation=45)
            plt.tight_layout()
            plt.show()

            # Print to separate the output of each document's plots
            print('\n\n')

    return final_results

```

# Analisi sentimentale

Per la visualizzazione dei risultati dell'analisi sentimentale è stato scelto un hexbin plot.

Per quanto riguarda i df raggruppati per article e topic si produce solo l'hexbin, in quanto i due df contengono molti documenti e il plot prodotto è molto esplicativo della distribuzione sentimentale.

Invece nei df raggruppati per issue e year, che producono un numero ristretto di documenti, anche se molto lunghi, l'hexbin plot era fortemente influenzato dall'implementazione della logica di chunk dei documenti e della media che ne deriva, per cui è stato affiancato da dei lineplot che mostrano la reale distribuzione degli score ottenuti dall'analisi sentimentale nei vari documenti.

```python
import numpy as np
import matplotlib.pyplot as plt

# Sentiment mapping
sentiment_mapping = {
    "Very Negative": 0,
    "Negative": 1,
    "Neutral": 2,
    "Positive": 3,
    "Very Positive": 4
}

def get_hexbin_sentiment_plot (sentiment_analysis):
  # Flatten the list of lists and extract sentiment & score
  x_values = []
  y_values = []

  for sublist in sentiment_analysis:  # Iterate over each inner list
      for item in sublist:  # Iterate over each dictionary inside the list
          sentiment = item.get("sentiment")
          score = item.get("score")

          if sentiment in sentiment_mapping and isinstance(score, (int, float)):
              x_values.append(sentiment_mapping[sentiment])
              y_values.append(score)

  # Create the hexbin plot
  plt.figure(figsize=(8, 6))
  hb = plt.hexbin(x_values, y_values, gridsize=10, cmap='YlGnBu', mincnt=1, extent=(0, 4, 0, 1))

  # Add colorbar
  plt.colorbar(label='Density')

  # Add labels and title
  plt.xlabel('Sentiment')
  plt.ylabel('Score')
  plt.title('Sentiment Analysis Hexbin Plot')

  # Adjust x-axis ticks
  plt.xticks(np.arange(5), ['Very Negative', 'Negative', 'Neutral', 'Positive', 'Very Positive'], rotation=45)

  # Show plot
  plt.show()
```

```python
print('Grouped by article:')
grby_art_sentiments =  process_sentiment_analysis(grouped_by_article['text'].tolist())
get_hexbin_sentiment_plot(grby_art_sentiments)
```

```python
print('Grouped by topic:')
grby_tp_sentiments = process_sentiment_analysis(grouped_by_topic['text'].tolist())
get_hexbin_sentiment_plot(grby_tp_sentiments)
```
